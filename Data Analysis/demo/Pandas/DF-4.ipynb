```python
import pandas as pd
```

```python
df1 = pd.DataFrame({ 'ID':[101, 102, 103, 104],
                     'NAME':['Nilesh', 'Ram', 'Sahil', 'Shyam'],
                     'AGE':[30, 45, 23, 25]})
df1
```

```python
df2 = pd.DataFrame({'ID':[105, 106, 107],
                     'NAME':['Rahul', 'Nitin', 'Prasad'],
                     'AGE':[35, 22, 32]})
df2
```

```python
pd.concat([df1, df2])
```

```python
pd.concat([df1, df2], axis=1)
```

```python
df3 = pd.DataFrame({'ID':[105, 106, 107],
                     'NAME':['Rahul', 'Nitin', 'Prasad'],
                     'SAL':[35, 22, 32]})
df3
```

```python
pd.concat([df1, df3])
```

```python
df4 = pd.DataFrame({'ADD':['Pune', 'Mumbai', 'Chennai', 'Kashmir'],
                     'DEPT':['DEV', 'DA', 'DS', 'AIML'],
                     'SAL':[35000, 22000, 32000, 55000]})
df4
```

```python
pd.concat([df1, df4], axis=1)
```

```python
emp = pd.DataFrame({ 'ID':[101, 102, 103, 104],
                     'NAME':['Nilesh', 'Ram', 'Sahil', 'Shyam'],
                     'AGE':[30, 45, 23, 25],
                   'DID': [1, 3, 1, 2]})
emp
```

```python
dept = pd.DataFrame({'DID': [1, 2, 3, 4],
                     'DNAME': ['IT', 'AI', 'ML', 'Dev']})
dept
```

```python
pd.merge(emp, dept)
```

```python
dept2 = pd.DataFrame({'ID': [1, 2, 3, 5],
                     'DNAME': ['IT', 'AI', 'ML', 'Dev']})
dept2
```

```python
pd.merge(emp, dept2, left_on='DID', right_on='ID')
```

```python
pd.merge(emp, dept, how='left')
```

```python
pd.merge(emp, dept, how='right')
```

```python
pd.merge(emp, dept, how='outer')
```
